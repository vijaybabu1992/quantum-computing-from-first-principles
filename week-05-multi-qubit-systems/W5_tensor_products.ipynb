```python
import numpy as np
```

```python
# Single Qubit Basis States
ket_0 = np.array([1,0])
ket_1 = np.array([0,1])
print("|0> = ", ket_0)
print("|1> = ", ket_1)
```

```python
# Two Qubit Basis via Tensor Product
ket_00 = np.kron(ket_0,ket_0)
ket_01 = np.kron(ket_0, ket_1)
ket_10 = np.kron(ket_1, ket_0)
ket_11 = np.kron(ket_1, ket_1)
print("|00> = ", ket_00)
print("|01> = ", ket_01)
print("|10> = ", ket_10)
print("|11> = ", ket_11)
```

```python
# Check Dimentions
print('Dimention of |00>:', ket_00.shape)
```

```python
# Build a separate Two-Qubit State

## Superposition on first qubit, |0> second
psi_1 = (ket_0 + ket_1)/np.sqrt(2)
psi_2 = ket_0

two_qubit_state = np.kron(psi_1, psi_2)
print("Sum of Probabilities:", np.sum(two_qubit_state**2))
```

```python
# Inerpretation
# Probabilities of each basis state
probabilities = two_qubit_state**2

labels = ["|00>","|01>","|10>","|11>"] 
for label, prob in zip(labels, probabilities):
    print(label,":", prob)
```
